```

import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt

# --- Functions ---
def pmt(principal, annual_rate, years, payments_per_year=12):
    r = annual_rate / payments_per_year
    n = years * payments_per_year
    if r == 0:
        return principal / n
    return principal * (r * (1 + r)**n) / ((1 + r)**n - 1)

def loan_schedule(principal, annual_rate, years, payments_per_year=12, extra_payment=0):
    payment = pmt(principal, annual_rate, years, payments_per_year)
    n = years * payments_per_year
    balance = principal
    schedule = []
    for month in range(1, n + 1):
        interest = balance * (annual_rate / payments_per_year)
        principal_paid = payment - interest + extra_payment
        if principal_paid > balance:
            principal_paid = balance
            total_payment = interest + principal_paid
        else:
            total_payment = payment + extra_payment
        balance -= principal_paid
        if balance < 0:
            balance = 0
        schedule.append([month, round(total_payment,2), round(principal_paid,2), round(interest,2), round(balance,2)])
        if balance == 0:
            break
    df = pd.DataFrame(schedule, columns=["Month","Payment","Principal","Interest","Balance"])
    return df

# --- Streamlit UI ---
st.title("💰 Loan Amortization Calculator")

principal = st.number_input("Loan Amount", min_value=1.0, value=100000.0)
annual_rate = st.number_input("Annual Interest Rate (%)", min_value=0.0, value=5.0) / 100
years = st.number_input("Loan Term (Years)", min_value=1, value=15)
extra_payment = st.number_input("Extra Monthly Payment", min_value=0.0, value=0.0)

if st.button("Calculate"):
    df = loan_schedule(principal, annual_rate, years, extra_payment=extra_payment)
    st.subheader("Monthly Payment")
    st.write(round(pmt(principal, annual_rate, years) + extra_payment, 2))

    st.subheader("Amortization Schedule (first 10 rows)")
    st.dataframe(df.head(10))

    st.subheader("Loan Balance Over Time")
    fig, ax = plt.subplots()
    ax.plot(df["Month"], df["Balance"], marker='o', color='green')
    ax.set_xlabel("Month")
    ax.set_ylabel("Remaining Balance")
    ax.grid(True)
    st.pyplot(fig)

    if st.button("Save Full Schedule to CSV"):
        df.to_csv("loan_schedule.csv", index=False)
        st.success("Saved as loan_schedule.csv")
```
