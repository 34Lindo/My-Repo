```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python

def pmt(principal, annual_rate, years, payments_per_year=12):
    """
    Calculate fixed payment (PMT) for a loan.
    """
    r = annual_rate / payments_per_year
    n = years * payments_per_year
    if r == 0:
        return principal / n
    payment = principal * (r * (1 + r)**n) / ((1 + r)**n - 1)
    return payment

def loan_schedule(principal, annual_rate, years, payments_per_year=12):
    """
    Create amortization schedule as a pandas DataFrame.
    """
    payment = pmt(principal, annual_rate, years, payments_per_year)
    n = years * payments_per_year
    balance = principal
    schedule = []

    for month in range(1, n + 1):
        interest = balance * (annual_rate / payments_per_year)
        principal_paid = payment - interest
        balance -= principal_paid
        if balance < 0:
            balance = 0
        schedule.append([month, round(payment, 2), round(principal_paid, 2), round(interest, 2), round(balance, 2)])
        if balance == 0:
            break

    df = pd.DataFrame(schedule, columns=["Month", "Payment", "Principal", "Interest", "Balance"])
    return df
```

```python

# Loan parameters
principal = 10000   # loan amount
annual_rate = 0.5   # 5% annual interest
years = 3         # term in years

# Generate schedule
df = loan_schedule(principal, annual_rate, years)
df.head(10)  # show first 10 months
```

```python

# Plot balance
plt.figure(figsize=(10,5))
plt.plot(df["Month"], df["Balance"], color="blue", linewidth=2)
plt.title("Loan Balance Over Time")
plt.xlabel("Month")
plt.ylabel("Remaining Balance")
plt.grid(True)
plt.show()
```
